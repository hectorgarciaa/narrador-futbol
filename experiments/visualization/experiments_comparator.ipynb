```python
import json

from football_ai.evaluation import ExperimentVisualizer
from football_ai.core import get_config
```

```python
config = get_config()
```

```python
TRACKS_PATH = str(config.get_path('paths', 'output', 'tracks_json')) + "/tracker/tracks3.json"
```

```python
with open(TRACKS_PATH, "r", encoding="utf-8") as f:
    tracks = json.load(f)
```

```python
ev = ExperimentVisualizer(tracks, ["player"])
```

```python
ev.show_bars_of_experiments("player")
```

```python
df_filtered = ev.create_df_summary("player")
df_filtered = df_filtered[df_filtered["n_tracks"] < 40]
df_filtered = df_filtered[df_filtered["max_speed"] < 250]
df_filtered = df_filtered[df_filtered["mean_team_flip_rate"] < 0.15]
ev.show_bars_of_experiments(None, df_filtered)
```
