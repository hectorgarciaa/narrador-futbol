```python
import json

from football_ai.evaluation import TrackVisualizer
from football_ai.core import get_config
```

```python
config = get_config()
```

```python
TRACKS_PATH = str(config.get_path('paths', 'output', 'tracks_json')) + "/tracker/tracks_m.json"
```

```python
with open(TRACKS_PATH, "r", encoding="utf-8") as f:
    tracks = json.load(f)
```

```python
tv = TrackVisualizer(tracks, "player")
```

```python
tv.show_all_hists_metrics_list()
```

```python
df_filtered = tv.get_df_metrics_list()
df_filtered = df_filtered["speed_values"].explode()
df_filtered = df_filtered[df_filtered > 15]
tv.show_hist(df_filtered, "speed_values")
```

```python
tv.show_tracks_evolution("confs", 1, 5, True)
```
